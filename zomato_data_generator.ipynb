```python
# ----------------------------------------
# Generate Fake Customer Data for Zomato Project
# ----------------------------------------

# Import required library
import pandas as pd
import random
from faker import Faker

# Create Faker object for Indian data
fake = Faker("en_IN")

# List of cuisine types
cuisines = [
    "North Indian",
    "South Indian",
    "Chinese",
    "Italian",
    "Mexican",
    "Fast Food",
    "Biryani",
    "Desserts"
]

# Empty list to store customer records
customers = []

# Generate data for 100 customers
for i in range(1, 101):

    customers.append({

        # Primary Key
        "customer_id": i,

        # Customer full name
        "name": fake.name(),

        # Customer email
        "email": fake.email(),

        # Customer phone number
        "phone": fake.phone_number(),

        # Customer address/location
        "location": fake.address().replace("\n", ", "),

        # Signup date
        "signup_date": fake.date_between(start_date="-3y", end_date="today"),

        # Premium membership (True or False)
        "is_premium": random.choice([True, False]),

        # Preferred cuisine
        "preferred_cuisine": random.choice(cuisines),

        # Average rating given by customer
        "average_rating": round(random.uniform(3.0, 5.0), 1),

        # Total orders placed
        "total_orders": random.randint(1, 300)
    })

# Convert list to DataFrame
customers_df = pd.DataFrame(customers)

# Display first 5 rows
customers_df.head()
```

```python
# ----------------------------------------
# Generate Fake Restaurant Data
# ----------------------------------------

# Import required libraries
import pandas as pd
import random

# List of cuisine types
cuisines = [
    "North Indian",
    "South Indian",
    "Chinese",
    "Italian",
    "Mexican",
    "Fast Food",
    "Biryani",
    "Desserts"
]

# Create an empty list to store restaurant records
restaurants = []

# Generate data for 50 restaurants
for i in range(1, 51):

    restaurants.append({

        # Primary Key
        "restaurant_id": i,

        # Restaurant Name
        "name": fake.company(),

        # Cuisine Type
        "cuisine_type": random.choice(cuisines),

        # Restaurant Location
        "location": fake.city(),

        # Restaurant Owner Name
        "owner_name": fake.name(),

        # Average Delivery Time (Minutes)
        "average_delivery_time": random.randint(20, 60),

        # Restaurant Contact Number
        "contact_number": fake.phone_number(),

        # Restaurant Rating (1–5)
        "rating": round(random.uniform(3.0, 5.0), 1),

        # Total Orders Received
        "total_orders": random.randint(100, 5000),

        # Restaurant Active Status
        "is_active": random.choice([True, False])
    })

# Convert the list into a Pandas DataFrame
restaurants_df = pd.DataFrame(restaurants)

# Display the first 5 rows
restaurants_df.head()
```

```python
# ----------------------------------------
# Generate Fake Orders Data
# ----------------------------------------

# Import required libraries
import pandas as pd
import random
from datetime import timedelta

# List of Payment Modes
payment_modes = [
    "Cash",
    "UPI",
    "Credit Card",
    "Debit Card"
]

# Create an empty list to store order records
orders = []

# Generate data for 500 orders
for i in range(1, 501):

    # Generate Order Date & Time
    order_date = fake.date_time_between(
        start_date="-1y",
        end_date="now"
    )

    # Generate Delivery Time (20–90 minutes after order)
    delivery_time = order_date + timedelta(
        minutes=random.randint(20, 90)
    )

    orders.append({

        # Primary Key
        "order_id": i,

        # Foreign Key - Customer
        "customer_id": random.randint(1, 100),

        # Foreign Key - Restaurant
        "restaurant_id": random.randint(1, 50),

        # Order Date
        "order_date": order_date,

        # Delivery Time
        "delivery_time": delivery_time,

        # Order Status (Weighted Probability)
        "status": random.choices(
            ["Pending", "Delivered", "Cancelled"],
            weights=[10, 80, 10],
            k=1
        )[0],

        # Total Bill Amount
        "total_amount": round(
            random.uniform(150, 2500), 2
        ),

        # Payment Mode
        "payment_mode": random.choice(payment_modes),

        # Discount Applied
        "discount_applied": round(
            random.uniform(0, 500), 2
        ),

        # Customer Feedback Rating
        "feedback_rating": random.randint(1, 5)

    })

# Convert list into DataFrame
orders_df = pd.DataFrame(orders)

# Display first 5 records
orders_df.head()
```

```python
# Query 29: Display delivery person's name and delivery status

cursor.execute("""
SELECT
    dp.name AS delivery_person_name,
    d.delivery_status
FROM delivery_persons dp
JOIN deliveries d
ON dp.delivery_person_id = d.delivery_person_id;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Query 30: Display complete order report by joining all five tables

cursor.execute("""
SELECT
    c.name AS customer_name,
    r.name AS restaurant_name,
    o.total_amount,
    o.status,
    dp.name AS delivery_person_name,
    d.delivery_status
FROM orders o
JOIN customers c
ON o.customer_id = c.customer_id
JOIN restaurants r
ON o.restaurant_id = r.restaurant_id
JOIN deliveries d
ON o.order_id = d.order_id
JOIN delivery_persons dp
ON d.delivery_person_id = dp.delivery_person_id;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
#  Find the locations with the highest number of orders

cursor.execute("""
SELECT
    c.location,
    COUNT(o.order_id) AS total_orders
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.location
ORDER BY total_orders DESC;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
#  Find the location with the highest number of cancelled orders

cursor.execute("""
SELECT
    c.location,
    COUNT(o.order_id) AS cancelled_orders
FROM customers c
JOIN orders o
ON c.customer_id = o.customer_id
WHERE o.status = 'Cancelled'
GROUP BY c.location
ORDER BY cancelled_orders DESC;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Find the restaurant with the highest number of orders

cursor.execute("""
SELECT
    r.name AS restaurant_name,
    COUNT(o.order_id) AS total_orders
FROM restaurants r
JOIN orders o
ON r.restaurant_id = o.restaurant_id
GROUP BY r.restaurant_id, r.name
ORDER BY total_orders DESC;
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
# Delivery class
class Delivery:

    # Constructor
    def __init__(self, db):

        # Receive Database object
        self.db = db

    # Create (Insert) Delivery
    def add_delivery(self, values):

        query = """
        INSERT INTO deliveries
        (delivery_id, order_id, delivery_person_id,
        delivery_status, distance, delivery_time,
        estimated_time, delivery_fee, vehicle_type)

        VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s)
        """

        self.db.execute_query(query, values)

        print("Delivery Added Successfully!")

    # Read (View) Deliveries
    def view_deliveries(self):

        query = "SELECT * FROM deliveries"

        return self.db.fetch_data(query)

    # Update Delivery Status
    def update_delivery_status(self, delivery_id, status):

        query = """
        UPDATE deliveries
        SET delivery_status=%s
        WHERE delivery_id=%s
        """

        self.db.execute_query(query, (status, delivery_id))

        print("Delivery Status Updated Successfully!")

    # Delete Delivery
    def delete_delivery(self, delivery_id):

        query = """
        DELETE FROM deliveries
        WHERE delivery_id=%s
        """

        self.db.execute_query(query, (delivery_id,))

        print("Delivery Deleted Successfully!")
```

```python
# Delivery Person class
class DeliveryPerson:

    # Constructor
    def __init__(self, db):

        # Receive Database object
        self.db = db

    # Create (Insert) Delivery Person
    def add_delivery_person(self, values):

        query = """
        INSERT INTO delivery_persons
        (delivery_person_id, name, contact_number,
        vehicle_type, total_deliveries,
        average_rating, location)

        VALUES (%s,%s,%s,%s,%s,%s,%s)
        """

        self.db.execute_query(query, values)

        print("Delivery Person Added Successfully!")

    # Read (View) Delivery Persons
    def view_delivery_persons(self):

        query = "SELECT * FROM delivery_persons"

        return self.db.fetch_data(query)

    # Update Delivery Person Location
    def update_location(self, delivery_person_id, location):

        query = """
        UPDATE delivery_persons
        SET location=%s
        WHERE delivery_person_id=%s
        """

        self.db.execute_query(query, (location, delivery_person_id))

        print("Location Updated Successfully!")

    # Delete Delivery Person
    def delete_delivery_person(self, delivery_person_id):

        query = """
        DELETE FROM delivery_persons
        WHERE delivery_person_id=%s
        """

        self.db.execute_query(query, (delivery_person_id,))

        print("Delivery Person Deleted Successfully!")
```

```python
import mysql.connector

# Create a Database class
class Database:

    # Constructor: automatically runs when an object is created
    def __init__(self):

        # Connect to the MySQL database
        self.conn = mysql.connector.connect(
            host="localhost",
            user="root",
            password="sujeet@1234",
            database="zomato_project"
        )

        # Create a cursor object to execute SQL queries
        self.cursor = self.conn.cursor()
        print("Database Connected Successfully!")
        

    # Method to execute INSERT, UPDATE, and DELETE queries
    def execute_query(self, query, values=None):

        # Execute query with values if provided
        if values:
            self.cursor.execute(query, values)

        # Execute query without values
        else:
            self.cursor.execute(query)

        # Save the changes to the database
        self.conn.commit()

    # Method to execute SELECT queries and fetch data
    def fetch_data(self, query):

        # Execute the SELECT query
        self.cursor.execute(query)

        # Return all rows from the query result
        return self.cursor.fetchall()
```

```python
# Execute INSERT, UPDATE and DELETE queries
def execute_query(self, query, values=None):

    try:
        # Execute query with values
        if values:
            self.cursor.execute(query, values)

        # Execute query without values
        else:
            self.cursor.execute(query)

        # Save changes permanently
        self.conn.commit()

        print("Query executed successfully!")

    # Handle database errors
    except mysql.connector.Error as err:

        print("Database Error:", err)

    # Handle any other unexpected errors
    except Exception as e:

        print("Error:", e)
```
